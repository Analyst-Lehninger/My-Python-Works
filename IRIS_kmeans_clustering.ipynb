```python
#STEP 01
#load the data and split into features and target variable
iris = load_iris()
X = iris.data
y = iris.target

feature_names = iris.feature_names
target_names = iris.target_names
print("Feature names:", feature_names)
print("Target names:", target_names)

print("/nType of X is:", type(X))
print("Shape of X is:", X.shape)
print("First 5 rows of X:\n", X[:5])
```

```python
#STEP 02
#split the data into training and testing sets

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.4, random_state = 1)

print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

```python
#step 03
#handling categorical data
#in this case, the target variable is already encoded as integers (0, 1, 2), so we can use it directly for modeling. If we had categorical features, we would need to encode them using techniques like one-hot encoding or label encoding.
#ML must use numbers, not text. So we need to convert categorical data into numerical format.
```

```python
#step 04
# training the model
from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression(max_iter = 200)
log_reg.fit(X_train, y_train)

print("Model training complete.")
```

```python
#step 05
# make predictions

y_pred = log_reg.predict(X_test)
print("Predictions:", y_pred)
```

```python
# step 06
# evaluate the model
from sklearn import metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
accuracy = accuracy_score(y_test, y_pred)

metrics.accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
print("Classification Report:\n", classification_report(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))

```

```python
#step 06 cont'd
#acuatlly test the model on new data

sample = [[3, 5, 4, 2], [2, 3, 5, 4]]
preds = log_reg.predict(sample)
pred_species = [iris.target_names[p] for p in preds]
print("Predictions:", pred_species)
```

## Step 07
Use K-Means clustering to see which iris measurements separate the species most clearly.

This section:
- standardizes the features,
- builds 3 clusters,
- ranks the features by cluster separation,
- and visualizes the strongest pair.

```python
# STEP 07
# Cluster the iris measurements and visualize the features that define the groups best

import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names
species_names = iris.target_names

# Standardize so all measurements are on the same scale
X_scaled = StandardScaler().fit_transform(X)

# Unsupervised clustering
kmeans = KMeans(n_clusters=3, random_state=42, n_init=20)
clusters = kmeans.fit_predict(X_scaled)

# Measure which features separate the clusters the most
feature_separation = np.ptp(kmeans.cluster_centers_, axis=0)
ranked_idx = np.argsort(feature_separation)[::-1]

top1, top2 = ranked_idx[:2]

print("Features ranked by cluster-separation strength:")
for i in ranked_idx:
    print(f"{feature_names[i]} -> {feature_separation[i]:.3f}")

print(f"\nBest two features for visualization: {feature_names[top1]} and {feature_names[top2]}")

# Cluster labels are arbitrary, so map them to the majority true species for a fair comparison
mapped_clusters = np.zeros_like(clusters)
cluster_species_map = {}
cluster_color_map = {}
for cluster_id in np.unique(clusters):
    mask = clusters == cluster_id
    majority_species = Counter(y[mask]).most_common(1)[0][0]
    mapped_clusters[mask] = majority_species
    cluster_species_map[cluster_id] = species_names[majority_species]
    cluster_color_map[cluster_id] = majority_species

agreement = accuracy_score(y, mapped_clusters)
print(f"Clustering agreement with the known species labels: {agreement:.2%}")

# PCA projection for a clean 2D overview using all 4 features
X_pca = PCA(n_components=2, random_state=42).fit_transform(X_scaled)

species_colors = ["#1f77b4", "#ff7f0e", "#2ca02c"]

fig, axes = plt.subplots(1, 3, figsize=(20, 5))

for species_id, species_name in enumerate(species_names):
    mask = y == species_id
    axes[0].scatter(
        X[mask, top1],
        X[mask, top2],
        label=species_name,
        color=species_colors[species_id],
        s=70,
        edgecolor="k",
        alpha=0.85,
    )
axes[0].set_title("Actual species")
axes[0].set_xlabel(feature_names[top1])
axes[0].set_ylabel(feature_names[top2])
axes[0].legend(title="Species")

for cluster_id in np.unique(clusters):
    mask = clusters == cluster_id
    species_id = cluster_color_map[cluster_id]
    axes[1].scatter(
        X[mask, top1],
        X[mask, top2],
        label=f"Cluster {cluster_id} ({cluster_species_map[cluster_id]})",
        color=species_colors[species_id],
        s=70,
        edgecolor="k",
        alpha=0.85,
    )
axes[1].set_title("K-Means clusters")
axes[1].set_xlabel(feature_names[top1])
axes[1].set_ylabel(feature_names[top2])
axes[1].legend(title="Cluster legend")

axes[2].barh(range(len(feature_names)), feature_separation[ranked_idx][::-1], color="teal")
axes[2].set_yticks(range(len(feature_names)))
axes[2].set_yticklabels([feature_names[i] for i in ranked_idx][::-1])
axes[2].set_title("Feature strength")
axes[2].set_xlabel("Cluster-center spread")

plt.suptitle("Iris clustering: which features separate the groups best?")
plt.tight_layout()
plt.show()

# PCA cluster plot with legend
plt.figure(figsize=(7, 5))
for cluster_id in np.unique(clusters):
    mask = clusters == cluster_id
    species_id = cluster_color_map[cluster_id]
    plt.scatter(
        X_pca[mask, 0],
        X_pca[mask, 1],
        label=f"Cluster {cluster_id} ({cluster_species_map[cluster_id]})",
        color=species_colors[species_id],
        s=70,
        edgecolor="k",
        alpha=0.85,
    )
plt.title("PCA view of the iris clusters")
plt.xlabel("Principal component 1")
plt.ylabel("Principal component 2")
plt.legend(title="PCA clusters")
plt.show()
```
